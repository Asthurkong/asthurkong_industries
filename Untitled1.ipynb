```python
from IPython.display import HTML

inicio = 10
fin = 30

# Definimos el código CSS en un bloque <style>
estilos_css = """
<style>
    .galeria {
        display: flex;
        flex-wrap: wrap;
        gap: 12px;
        padding: 15px;
        background-color: #f8f9fa;
        border-radius: 10px;
        box-shadow: inset 0 0 5px rgba(0,0,0,0.05);
    }
    .galeria img {
        border-radius: 8px;
        object-fit: cover;
        box-shadow: 0 2px 5px rgba(0,0,0,0.15);
        transition: transform 0.2s ease, box-shadow 0.2s ease;
        cursor: pointer;
    }
    .galeria img:hover {
        transform: scale(1.1);
        box-shadow: 0 5px 15px rgba(0,0,0,0.3);
    }
</style>
"""

# Inicializamos agregando los estilos y abriendo el contenedor div
miHTML = estilos_css + '<div class="galeria">'

for i in range(inicio, fin):
    # Nota: Se corrigió la etiqueta HTML de <image> a <img>
    miHTML += f'<img src="https://picsum.photos/id/{i}/80/80" alt="Imagen {i}">'

# Cerramos el contenedor div
miHTML += '</div>'

# Ejecutamos el código HTML
display(HTML(miHTML))
```
